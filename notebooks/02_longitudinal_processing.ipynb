# 02 - Longitudinal processing

Load the simulated cohort, validate it, and convert `memory`/`attention`/`language`
to demographically-adjusted z-scores (`src/tasks/norms.py`) so they're on the
same scale as the CLI test battery's normed subtests. If any real CLI sessions
have been recorded (`data/sessions/sessions_log.csv`), merge them in on the
shared `memory_z`/`attention_z`/`language_z` schema.

```python
import sys
sys.path.append("..")

import os
import pandas as pd

from src.data_ingest import load_simulated_longitudinal, load_real_sessions_normed
from src.preprocessing import CANONICAL_COLUMNS, merge_simulated_and_real
```

```python
sim_normed, domain_norms = load_simulated_longitudinal("../data/simulated/longitudinal_simulated.csv")
sim_normed["source"] = "simulated"
sim_normed.head()
```

```python
real_normed = load_real_sessions_normed(
    sessions_path="../data/sessions/sessions_log.csv",
    participants_path="../data/sessions/participants.csv",
)

if real_normed.empty:
    print("No real CLI sessions recorded yet - using the simulated cohort only.")
    combined = sim_normed[CANONICAL_COLUMNS]
else:
    print(f"Merging {len(real_normed)} real session rows with the simulated cohort.")
    combined = merge_simulated_and_real(sim_normed, real_normed)

combined.shape
```

```python
os.makedirs("../data/processed", exist_ok=True)
combined.to_csv("../data/processed/longitudinal_normed.csv", index=False)
combined.head()
```
